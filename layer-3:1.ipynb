Next, I'll define a function to read all CSV files within a specified subfolder, apply the loaded header, and combine them into a single DataFrame.

```python
def load_and_process_folder_data(folder_name, header):
    folder_path = os.path.join(os.getcwd(), folder_name)
    all_files = os.listdir(folder_path)
    csv_files = [f for f in all_files if f.endswith('.csv')]

    if not csv_files:
        print(f"No CSV files found in {folder_name}")
        return pd.DataFrame()

    list_dfs = []
    for csv_file in csv_files:
        file_path = os.path.join(folder_path, csv_file)
        try:
            # Read without header, then assign the predefined header
            df = pd.read_csv(file_path, header=None)
            if len(df.columns) == len(header):
                df.columns = header
                list_dfs.append(df)
            else:
                print(f"Skipping {csv_file}: column count mismatch with header ({len(df.columns)} vs {len(header)}).")
        except Exception as e:
            print(f"Error reading {csv_file}: {e}")

    if list_dfs:
        return pd.concat(list_dfs, ignore_index=True)
    else:
        return pd.DataFrame()
```

Now, let's aggregate the data for each location on a daily basis. Since each file represents a day, we'll convert the 'Date_time' column to datetime, extract the date, and then sum the total power consumed.

```python
# Function to process and aggregate data by day
def aggregate_by_day(df, name):
    if df.empty:
        print(f"\n{name} DataFrame is empty, skipping daily aggregation.")
        return pd.DataFrame()

    # Convert 'Date_time' to datetime, handling potential errors
    # The format seems to be 'DD-MM-YYYY HH:MM:SS'
    df['Date_time'] = pd.to_datetime(df['Date_time'], format='%d-%m-%Y %H:%M:%S', errors='coerce')

    # Drop rows where Date_time conversion failed
    df.dropna(subset=['Date_time'], inplace=True)

    if df.empty:
        print(f"\n{name} DataFrame became empty after dropping invalid dates, skipping daily aggregation.")
        return pd.DataFrame()

    df['Date'] = df['Date_time'].dt.date

    # Aggregate by Date and sum relevant numerical columns
    # Selecting a few key columns to demonstrate daily aggregation
    daily_agg_df = df.groupby('Date').agg(
        Daily_kW_Total=('kW_Total', 'sum'),
        Daily_kwh_delivered_calculated=('kwh deliverd calculated', 'sum')
    ).reset_index()

    print(f"\n{name} Daily Aggregated Data - First 5 rows:")
    display(daily_agg_df.head())
    print(f"{name} Daily Aggregated Data Shape: {daily_agg_df.shape}")

    return daily_agg_df

# Aggregate Hostel data by day
hostel_daily_df = aggregate_by_day(hostel_df.copy(), 'Hostel')

# Aggregate College data by day
college_daily_df = aggregate_by_day(college_df.copy(), 'College')

# Aggregate Hospital data by day
hospital_daily_df = aggregate_by_day(hospital_df.copy(), 'Hospital')
```

## Phase 3 — build features (preprocessing)

Real readings come in every ~2 minutes, not daily, so before the 7-day rolling slope from the prototype notebook (§3) makes sense we resample each machine to one row per day. Steps: filter to the dominant device per location, coerce pipeline columns to numeric, resample to daily means, compute phase imbalance %, normalize the harmonics/neutral current by the load indicator, then run the prototype's exact `rolling_slope` / `add_trend_features` against it.

```python
pipeline_cols = ['kW_Total', 'PF_Ave', 'Current_Total', 'Current_R_phase', 'Current_Y_phase',
                  'Current_B_phase', 'Current_R_Harm', 'Current_Y_Harm', 'Current_B_Harm', 'Neutral_current']

def prep_machine_daily(df, device_id, machine_id):
    d = df[df['Deviced_id'] == device_id].copy()
    d['Date_time'] = pd.to_datetime(d['Date_time'], format='%d-%m-%Y %H:%M:%S', errors='coerce')
    for col in pipeline_cols:
        d[col] = pd.to_numeric(d[col], errors='coerce')
    d = d.dropna(subset=['Date_time']).sort_values('Date_time')
    daily = d.set_index('Date_time')[pipeline_cols].resample('D').mean()
    daily['rows_that_day'] = d.set_index('Date_time')['kW_Total'].resample('D').count()
    daily['machine_id'] = machine_id
    return daily.reset_index().rename(columns={'Date_time': 'date'})

daily_hostel = prep_machine_daily(hostel_df, 1, 'hostel_J4')
daily_college = prep_machine_daily(college_df, 23, 'college_AV11')
daily_hospital = prep_machine_daily(hospital_df, 3.0, 'hospital_AV15')

daily_all = pd.concat([daily_hostel, daily_college, daily_hospital]).reset_index(drop=True)
print(daily_all.shape)
print(daily_all.groupby('machine_id')['date'].agg(['min', 'max', 'count']))
daily_all.head()
```

```python
# Phase imbalance % and normalization by the load indicator (Current_Total)
phases = daily_all[['Current_R_phase', 'Current_Y_phase', 'Current_B_phase']]
daily_all['phase_imbalance_pct'] = (
    (phases.max(axis=1) - phases.min(axis=1)) / phases.mean(axis=1) * 100
)

for col in ['Current_R_Harm', 'Current_Y_Harm', 'Current_B_Harm', 'Neutral_current']:
    daily_all[f'{col}_norm'] = daily_all[col] / daily_all['Current_Total']

daily_all[['machine_id', 'date', 'Current_Total', 'phase_imbalance_pct',
           'Current_R_Harm_norm', 'Neutral_current_norm']].head()
```

```python
# From the prototype notebook §3 (Predictive_Maintenance_Prototype.ipynb), unchanged logic —
# run against the real, resampled daily_all instead of synthetic data.
import numpy as np

def rolling_slope(series, window):
    slopes = []
    for i in range(len(series)):
        if i < window - 1:
            slopes.append(np.nan)
            continue
        y = series.iloc[i - window + 1:i + 1].values
        x = np.arange(window)
        slopes.append(np.polyfit(x, y, 1)[0])
    return pd.Series(slopes, index=series.index)

def add_trend_features(g, window=7):
    g = g.sort_values('date').copy()
    for col in ['Current_R_Harm_norm', 'Current_Y_Harm_norm', 'Current_B_Harm_norm',
                'PF_Ave', 'Neutral_current_norm', 'phase_imbalance_pct']:
        g[f'{col}_slope_{window}d'] = rolling_slope(g[col], window)
    return g

# loop per machine explicitly (see prototype note: groupby().apply() can silently drop
# the group key column in some pandas versions when the function returns the full frame)
daily_feat = pd.concat([add_trend_features(g) for _, g in daily_all.groupby('machine_id')]).reset_index(drop=True)

slope_cols = [c for c in daily_feat.columns if c.endswith('_slope_7d')]
print("non-null slope rows per machine:")
print(daily_feat.groupby('machine_id')[slope_cols[0]].apply(lambda s: s.notna().sum()))
daily_feat[['machine_id', 'date'] + slope_cols].dropna().head()
```

Spot-check: slope values are non-zero and vary day to day rather than sitting flat at zero or NaN — matches the "done when" bar for Phase 3.

## Phase 4 — Health Index (Stage A)

**Baseline note:** no manufacturer spec was available, so `BASELINE` below is estimated empirically from this fleet's own full-year history (mean/std per feature, across all three machines, since these are load-normalized ratios that should be comparable across differently-sized transformers). **Flag as "to confirm" — replace with manufacturer/OEM spec numbers if you get them**, per the TODO.

```python
# Population-prior baseline — TO CONFIRM. Estimated from this fleet's own history since no
# manufacturer spec was supplied (see prototype notebook §4 for the production-intent version,
# where this comes from Layer 0 config / Unit 3's Stage 1 priors instead).
baseline_source_cols = ['Current_R_Harm_norm', 'Current_Y_Harm_norm', 'Current_B_Harm_norm',
                         'PF_Ave', 'Neutral_current_norm', 'phase_imbalance_pct']
BASELINE = {col: (daily_all[col].mean(), daily_all[col].std()) for col in baseline_source_cols}
for col, (mean, std) in BASELINE.items():
    print(f"  {col}: mean={mean:.4f}, std={std:.4f}")

def compute_health_index(row):
    z_scores = [abs(row[col] - mean) / std for col, (mean, std) in BASELINE.items()]
    avg_z = np.mean(z_scores)
    return max(0, 100 - avg_z * 12)

daily_feat['health_index'] = daily_feat.apply(compute_health_index, axis=1)
daily_feat.groupby('machine_id')['health_index'].agg(['min', 'max', 'mean'])
```

## Phase 5 — early-warning detection (CUSUM)

```python
# From the prototype notebook §5, unchanged, run against each real machine's health_index series.
def cusum_flags(series, threshold=8, drift=0.5):
    s_pos, s_neg, flags = 0, 0, []
    baseline_mean = series.dropna().iloc[:10].mean()
    for val in series.fillna(method='ffill'):
        s_pos = max(0, s_pos + (baseline_mean - val) - drift)
        s_neg = min(0, s_neg + (baseline_mean - val) + drift)
        flags.append(s_pos > threshold or abs(s_neg) > threshold)
    return flags

for mid, g in daily_feat.groupby('machine_id'):
    g = g.sort_values('date')
    flags = cusum_flags(g['health_index'])
    n_flags = sum(flags)
    first_flag = next((i for i, f in enumerate(flags) if f), None)
    when = g['date'].iloc[first_flag] if first_flag is not None else 'never'
    print(f"{mid}: first change-point flagged at {when}  ({n_flags}/{len(flags)} days flagged total)")
```

```python
# Why is CUSUM over-flagging? Real health_index is noisier day-to-day than the prototype's
# smooth synthetic 60-day series, so default threshold=8/drift=0.5 (tuned for that smooth
# signal) trips almost immediately on real data.
for mid, g in daily_feat.groupby('machine_id'):
    g = g.sort_values('date')
    hi = g['health_index']
    print(f"{mid}: health_index day-to-day std={hi.diff().std():.2f}, "
          f"overall std={hi.std():.2f}, first-10-day baseline mean={hi.iloc[:10].mean():.2f}")
```

**Finding:** the flags are explainable, but not clean — day-to-day health-index std (8–26) is on the same order as `threshold=8`, so CUSUM (tuned in the prototype for a smooth 60-day synthetic series) trips almost every day on real, noisier data. Function runs without error, satisfying the letter of Phase 5's "done when," but the defaults need re-tuning (larger `threshold`/`drift`, or smoothing `health_index` first) before this is a useful early-warning signal rather than a constant alarm. **Flag as "to confirm"** alongside the Phase 4 baseline.

## Phase 6 — package the output & prep to demo

```python
# From the prototype notebook §7, adapted only for the real (normalized) feature column names.
import uuid

def predict_maintenance_risk(machine_id, latest_row, confidence_stage='stage_a_unsupervised'):
    health_index = latest_row['health_index']
    risk_tier = 'high' if health_index < 40 else 'medium' if health_index < 65 else 'low'
    return {
        "event_id": str(uuid.uuid4()),
        "machine_id": machine_id,
        "timestamp": str(latest_row['date']),
        "source_unit": "predictive_maintenance",
        "event_type": "maintenance_risk",
        "severity": "alert" if risk_tier == "high" else "info",
        "health_index": round(float(health_index), 1),
        "risk_tier": risk_tier,
        "confidence_stage": confidence_stage,
        "contributing_features": {
            "current_r_harm_norm_slope_7d": round(float(latest_row['Current_R_Harm_norm_slope_7d']), 4),
            "pf_ave_slope_7d": round(float(latest_row['PF_Ave_slope_7d']), 4),
            "phase_imbalance_slope_7d": round(float(latest_row['phase_imbalance_pct_slope_7d']), 3),
        },
    }

import json
for mid, g in daily_feat.groupby('machine_id'):
    latest = g.dropna(subset=['health_index', 'Current_R_Harm_norm_slope_7d']).sort_values('date').iloc[-1]
    result = predict_maintenance_risk(mid, latest)
    print(json.dumps(result, indent=2))
    print()
```

- **hospital_AV15 — low (86.4):** currents, PF, and phase balance all sit close to fleet-normal, with only a mild PF/imbalance drift.
- **college_AV11 — low (68.4)**, but closest to the medium boundary: its PF_Ave has been trending *up* recently (slope +0.11), pulling the index up from a much noisier, historically lower baseline (~48 average) — worth watching rather than treating as clearly healthy.

## Devices / sensor channels we have

The hospital's real device (`Deviced_id 3.0`, the 500 KVA transformer meter — 170,741 of 170,746 rows) is a **single multi-function power-quality meter**, not multiple physical devices. What looks like "many devices" in `header.csv` is really this one meter's internal measurement channels, each fed by its own sensor (CT/PT) or computed register. Grouped by channel, with real observed ranges from the hospital's own data pulled above:

| Channel (sensor) | Columns | Functionality | Why it matters for a hospital transformer | What we can predict / analyze |
|---|---|---|---|---|
| **Active power (kW) meter** | `kW_Total`, `kW_R/Y/B_phase` | Real power actually doing work, per phase and total. Observed: Total 0–356 kW, phases individually swing negative (–39 kW) even though Total floors at 0. | Hospitals run life-support, HVAC, sterilization, imaging — sustained overload directly threatens uptime. ## Data-quality flags found in this pass (worth fixing before Health Index trusts these channels)

Pulling real min/mean/max per channel surfaced issues beyond the ones already in Phase 2:
- **Dead registers**: `PF average delivered`, `Amps average Delivered`, `kWh_Delivered`, `VAh_Delivered`, `VARh_Ind./Cap._Delivered` are constant **0** for the entire year — not wired on this meter. `PF average Received` is constant ~-0.945 — effectively frozen, not a live signal either. None of these should be fed into the Health Index or trend features; the current pipeline already avoids them.
- **Frequency dropouts**: `Frequency` ranges from **0 to 63.5 Hz**, both physically implausible for a 50 Hz grid — these are almost certainly comms/sensor dropouts, not real events. Should be clipped to a sane band (e.g. 47–53 Hz) and treated as missing outside it.
- **Harmonic outliers**: `Current_R/Y/B_Harm` average 5–7% (plausible) but spike as high as **500–710%** — not physically real THD values. These extreme spikes should be capped or filtered (e.g. drop values above ~100%) before they're allowed to swing the Health Index or the rolling slope in Phase 3.

## What's still needed for a more complete Layer 3 health analysis

Everything above comes from one electrical meter. That's enough to prove the Stage A pipeline (Phases 1–6), but real transformer condition monitoring leans heavily on signals this meter simply doesn't have. Roughly in order of predictive value for a transformer specifically (vs. the load it serves):

**Highest-value gap — thermal/insulation condition:**
- **Top-oil and winding-hotspot temperature (OTI/WTI)** — the single most direct transformer health signal there is; almost every industry-standard predictive model for transformers is built around temperature rise vs. load, not electrical signature alone.
- **Dissolved Gas Analysis (DGA)**, even periodic lab samples rather than continuous online sensors — this is the gold-standard signal for detecting incipient insulation breakdown, arcing, and overheating *before* it becomes an electrical symptom. Nothing in the current dataset substitutes for this.
- **Oil level and oil dielectric strength / moisture (BDV/PPM tests)** — moisture ingress is a leading cause of insulation failure and isn't visible in any electrical channel here.

**Mechanical / auxiliary systems:**
- **Tap-changer (OLTC) position and operation count** — a high-wear mechanical component that fails independently of the windings; not observable electrically until it's already causing voltage problems.
- **Cooling system status** (fan/pump run state, cooling mode ONAN/ONAF) — without this, a temperature rise can't be distinguished from "normal, cooling just kicked in" vs. "cooling has failed."
- **Bushing monitoring** (capacitance/tan-delta, partial discharge) — bushings are a common hospital-transformer failure point and are invisible to the panel meter.
- **Breather/silica-gel condition** — cheap, simple, and a real leading indicator of moisture ingress that's currently not logged anywhere.
- **Vibration/acoustic sensors** — useful for core looseness or mechanical winding deformation, again invisible electrically until advanced.

**Context data (not new sensors, but needed to interpret what we already have):**
- **Nameplate rating and load history in % of rated capacity**, not just raw amps — `Current_Total` alone doesn't say how close to the transformer's actual limit the machine is running.
- **Ambient temperature / weather** — needed to separate "hot because it's summer" from "hot because it's degrading." Could be pulled from an external weather API keyed on `Date_time` rather than a new physical sensor.
- **Protection-relay trip/alarm event log** (Buchholz relay, differential/overcurrent trips) — this is exactly the real-world label source Stage B needs (per `Predictive_Maintenance_Implementation_Plan.md`), and it's currently the single biggest gap for turning this from an unsupervised pipeline into a validated one.
- **Maintenance/inspection log** — same gap as above; without at least 10-20 real logged events, there's no way to check whether the Health Index or CUSUM flags actually mean anything for *this* fleet.
- **Criticality weighting for the risk tiers** — this is a hospital: an identical Health Index reading here probably warrants a higher-urgency response than the same reading at the college. Worth a site-criticality multiplier once Layer 4/5 consume these `maintenance_risk` events, so alerting doesn't treat all three sites as equally urgent.
